# Drought – Temperatur, Niederschlag und De-Martonne-Index

Die Konturkarte verbindet die beiden Eingangsgrößen mit den Klassen des Ariditätsindex.
Datenquellen, Klassen und Grenzen der Interpretation sind in der README beschrieben.
Vor dem Ausführen die DWD-Jahresraster mit `python -m drought download` laden.


```python
from pathlib import Path
from IPython.display import Image, display
from drought.analysis import load_climate, quantile_breaks
from drought.plots import draw

data_dir = Path("data")
output_dir = Path("outputs/notebook")
start, end = 1996, 2025
```

## Periodenmittel

Nur Zellen mit vollständigen Daten für beide Größen gehen ein. Der Index wird aus
den Periodenmitteln berechnet, nicht durch Mittelung der jährlichen Indizes.


```python
climate = load_climate(data_dir, start, end)
p_breaks = quantile_breaks(climate.precipitation)
t_breaks = quantile_breaks(climate.temperature)
print("Quartilgrenzen Niederschlag (mm):", p_breaks)
print("Quartilgrenzen Temperatur (°C):", t_breaks)
```

## Bivariate Karte mit Klassengrenzen

Die Originalpalette verbindet Ocker und Türkis. Der Mehrheitsfilter glättet ausschließlich
die dMI-Klassen, deren Grenzen als Konturen erscheinen.


```python
path = output_dir / "contours.png"
draw(climate, path, "contours", p_breaks, t_breaks, f"{start}–{end} (fest)")
display(Image(filename=str(path)))
```

## Die beiden Ansichten getrennt

Die reine Indexkarte zeigt die gewählte Klassifikation. Ohne Konturen lassen sich
die bivariaten Farbflächen für sich betrachten.


```python
for kind in ("demartonne", "bivariate"):
    path = output_dir / f"{kind}.png"
    draw(climate, path, kind, p_breaks, t_breaks, f"{start}–{end} (fest)")
    display(Image(filename=str(path)))
```
